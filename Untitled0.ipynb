```python
# Import libraries and fetch the list of 133 supported languages
from deep_translator import GoogleTranslator
import pandas as pd
import os
import zlib
import time

supported_langs = GoogleTranslator().get_supported_languages(as_dict=True)
print(f"Total supported languages: {len(supported_langs)}")
```

```python
# Upload Federalist_Papers.txt and read it into input_text
from google.colab import files

uploaded = files.upload()
filename = list(uploaded.keys())[0]

with open(filename, 'r', encoding='utf-8') as f:
    input_text = f.read()

print(f"Loaded {filename}: {len(input_text)} characters")
```

```python
# Remove Gutenberg header + table of contents, keep only the essays
anchor_phrase = "To the People of the State of New York"
start_idx = input_text.find(anchor_phrase)
heading_idx = input_text.rfind("FEDERALIST No. 1", 0, start_idx)
clean_text = input_text[heading_idx if heading_idx != -1 else start_idx:].strip()

print(f"Cleaned text: {len(clean_text)} characters (from {len(input_text)} original)")
```

```python
# Split clean_text into ~4500-character chunks (Google Translate's request limit)
def chunk_text(text, max_chars=4500):
    sentences = text.replace('\n', ' \n').split('. ')
    chunks, current_chunk = [], ""
    for sentence in sentences:
        piece = sentence if sentence.endswith('.') else sentence + '. '
        if len(current_chunk) + len(piece) <= max_chars:
            current_chunk += piece
        else:
            if current_chunk:
                chunks.append(current_chunk.strip())
            current_chunk = piece
    if current_chunk:
        chunks.append(current_chunk.strip())
    return chunks

text_chunks = chunk_text(clean_text)
print(f"Split into {len(text_chunks)} chunks")
```

```python
# Translate all chunks into all 133 languages (parallel, with retry) and save each as a .txt file
from concurrent.futures import ThreadPoolExecutor, as_completed

output_folder = 'translated_files'
os.makedirs(output_folder, exist_ok=True)

def translate_chunk_with_retry(chunk, lang_code, max_retries=3):
    for attempt in range(max_retries):
        try:
            result = GoogleTranslator(source='en', target=lang_code).translate(chunk)
            return result if result else ""
        except Exception as e:
            if attempt < max_retries - 1:
                time.sleep(2 * (attempt + 1))
            else:
                raise e

def translate_one_language(lang_name, lang_code):
    try:
        translated_chunks = [translate_chunk_with_retry(c, lang_code) for c in text_chunks]
        full_translated_text = " ".join(translated_chunks)
        file_size = len(full_translated_text.encode('utf-8'))
        compressed_size = len(zlib.compress(full_translated_text.encode('utf-8')))

        safe_name = lang_name.replace(' ', '_').replace('/', '-')
        with open(os.path.join(output_folder, f"{safe_name}.txt"), 'w', encoding='utf-8') as f:
            f.write(full_translated_text)

        return {'language': lang_name, 'language_code': lang_code, 'translated_text': full_translated_text,
                'char_count': len(full_translated_text), 'file_size': file_size,
                'compressed_size': compressed_size, 'status': 'success'}
    except Exception as e:
        return {'language': lang_name, 'language_code': lang_code, 'translated_text': None,
                'char_count': None, 'file_size': None, 'compressed_size': None,
                'status': f'failed: {str(e)[:100]}'}

results = []
lang_items = list(supported_langs.items())

with ThreadPoolExecutor(max_workers=4) as executor:
    futures = {executor.submit(translate_one_language, name, code): name for name, code in lang_items}
    for i, future in enumerate(as_completed(futures), 1):
        result = future.result()
        results.append(result)
        icon = "✅" if result['status'] == 'success' else "❌"
        print(f"[{i}/{len(lang_items)}] {icon} {result['language']}")
        if i % 10 == 0:
            pd.DataFrame(results).to_csv('translated_output.csv', index=False)

translation_df = pd.DataFrame(results)
translation_df.to_csv('translated_output.csv', index=False)
print(f"\nDone: {(translation_df['status']=='success').sum()} / {len(lang_items)} successful")
```

```python
# Add compression_ratio column and show stats for all 133 languages
original_file_size = len(clean_text.encode('utf-8'))
original_compressed_size = len(zlib.compress(clean_text.encode('utf-8')))
original_char_count = len(clean_text)

translation_df['compression_ratio'] = translation_df['compressed_size'] / translation_df['file_size']
translation_df.to_csv('translated_output.csv', index=False)

pd.set_option('display.max_rows', 140)
print(f"English baseline — chars: {original_char_count}, compression ratio: {original_compressed_size/original_file_size:.3f}\n")
print(translation_df.sort_values('char_count')[['language','char_count','compression_ratio']].to_string(index=False))
```

```python
# Compute word count, avg characters/word, and unique characters per language
os.makedirs('character_analysis', exist_ok=True)

stats_rows = []
for _, row in translation_df.iterrows():
    if row['status'] != 'success':
        continue
    text = row['translated_text']
    word_count = len(text.split())
    unique_chars = len(set(text.replace(' ', '').replace('\n', '')))
    stats_rows.append({
        'language': row['language'], 'char_count': row['char_count'], 'word_count': word_count,
        'avg_characters_per_word': round(row['char_count'] / word_count, 3) if word_count else 0,
        'unique_characters': unique_chars, 'file_size': row['file_size'],
        'compressed_size': row['compressed_size'], 'compression_ratio': row['compression_ratio']
    })

char_stats_df = pd.DataFrame(stats_rows).sort_values('char_count').reset_index(drop=True)
char_stats_df.index += 1
char_stats_df.to_csv('character_analysis/character_statistics.csv', index=True, index_label='rank')
print(f"Saved character_statistics.csv — {len(char_stats_df)} languages ranked by char_count")
```

```python
# Chart: all 133 languages ranked by character count (green=efficient, red=inefficient)
import matplotlib.pyplot as plt
import matplotlib.cm as cm

fig, ax = plt.subplots(figsize=(10, 28))
norm = plt.Normalize(char_stats_df['char_count'].min(), char_stats_df['char_count'].max())
colors = cm.RdYlGn_r(norm(char_stats_df['char_count']))

ax.barh(char_stats_df['language'], char_stats_df['char_count'], color=colors)
ax.invert_yaxis()
ax.set_xlabel('Character Count')
ax.set_title('All 133 Languages Ranked by Character Efficiency (Federalist Papers)')
ax.axvline(original_char_count, color='blue', linestyle='--', label=f'English source ({original_char_count} chars)')
ax.legend(loc='lower right')
plt.tight_layout()
plt.savefig('character_analysis/full_133_ranking_chart.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Rename saved .txt files to include language_code, and save a name<->code lookup
for _, row in translation_df.iterrows():
    if row['status'] != 'success' or row['language'] == 'english (source)':
        continue
    safe_name = row['language'].replace(' ', '_').replace('/', '-')
    old_path = os.path.join('translated_files', f"{safe_name}.txt")
    new_path = os.path.join('translated_files', f"{safe_name}_{row['language_code']}.txt")
    if os.path.exists(old_path):
        os.rename(old_path, new_path)

translation_df[['language', 'language_code']].to_csv('translated_files/language_code_lookup.csv', index=False)
print("Renamed files to include language_code (e.g. japanese_ja.txt) + saved lookup table")
```
